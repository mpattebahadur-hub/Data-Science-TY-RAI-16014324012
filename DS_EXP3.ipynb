FIND THE MEAN OF MAXIMUM TEMPERATURE RECORDED

```python
data_Max = df["MaxT"].dropna().tolist()

total = 0

for x in data_Max:
    total = total + x

n = len(data_Max)
mean = total / n

print("Mean of Maximum Temperature =", mean)
```

The average maximum temperature recorded in the dataset is approximately 32.06 degree C, indicating that the study area generally experiences a warm climate.

FIND THE MEDIAN OF RELATIVE HUMIDITY IN THE MORNING

```python
data_Humid = df["RH1"].dropna().tolist()
n = len(data_Humid)

# Bubble Sort
for i in range(n):
    for j in range(0, n - i - 1):
        if data_Humid[j] > data_Humid[j + 1]:
            temp = data_Humid[j]
            data_Humid[j] = data_Humid[j + 1]
            data_Humid[j + 1] = temp


if n % 2 == 0:
    median = (data_Humid[n // 2 - 1] + data_Humid[n // 2]) / 2
else:
    median = data_Humid[n // 2]
print("Median of Relative Humidity in Morning =", median)
```

The median morning relative humidity is 87%, which means that half of the recorded RH1 values are below 87% and the other half are above it. This indicates that the mornings are generally associated with high relative humidity.

FIND THE MODE OF RELATIVE HUMIDITY IN AFTERNOON

```python
data_HumidA = df["RH2"].dropna().tolist()

frequency = {}
for x in data_HumidA:
    if x in frequency:
        frequency[x] = frequency[x] + 1
    else:
        frequency[x] = 1

max_count = 0
mode = None

for x in frequency:

    if frequency[x] > max_count:

        max_count = frequency[x]
        mode = x


print("Mode of Relative Humidity in Afternoon =", mode)
```

The most frequently occurring afternoon relative humidity is 29%. This suggests that relatively low humidity conditions occur frequently during the afternoon.




VARIANCE OF THE RADIATION

```python
data_Rad = df["Radiation"].dropna().tolist()

n = len(data_Rad)
total = 0

for x in data_Rad:
    total = total + x

mean_rad = total / n


# Calculate variance
sum_squared_difference = 0

for x in data_Rad:
    difference = x - mean_rad
    sum_squared_difference = (sum_squared_difference + difference * difference)

variance = sum_squared_difference /n
print("Variance of Radiation =", variance)
```

The variance of radiation is 20.32, indicating that the radiation values are spread around their mean rather than remaining constant. This variation can occur due to changes in weather conditions, cloud cover, seasons, and daily atmospheric conditions.

STANDARD DEVIATION

```python
Std= np.sqrt(variance)
print(mean_rad)
print("Standard Deviation of Radiation =", Std)
```

The standard deviation of approximately 4.51 indicates the typical deviation of radiation values from the mean radiation value of 17.88. Therefore, radiation measurements show a moderate level of variability and are not concentrated at exactly one value.

INTERQUARTILE RANGE AND OUTLIERS

```python
attribute = "MaxT"
data = df[attribute].dropna()
# Calculate Quartiles
Q1 = data.quantile(0.25)
Q2 = data.quantile(0.50)
Q3 = data.quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1
print("Q1 =", Q1)
print("Q2 =", Q2)
print("Q3 =", Q3)
print("IQR =", IQR)

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("\nLower Bound =", lower_bound)
print("Upper Bound =", upper_bound)

outliers = data[(data < lower_bound) | (data > upper_bound)]

print("Number of Outliers =", outliers.count())

print("Outlier Values:")
print(outliers.tolist())

print("Minimum Extreme Value =", data.min())
print("Maximum Extreme Value =", data.max())
```

Most of the maximum temperature values lie between 29 degrees and 35 degrees, with an IQR of 6 degrees. There are 4 low-temperature outliers: 17.8 degrees, 16.5 degrees, 19 degrees, and 19.5 degrees, as they fall below the lower bound of 20 degrees. There are no high-temperature outliers, since the maximum value of 43.5 degrees is within the upper bound of 44 degrees.
